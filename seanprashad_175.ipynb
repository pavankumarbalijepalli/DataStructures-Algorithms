```python
# https://leetcode.com/problems/contains-duplicate/

def containsDuplicate(nums: List[int]) -> bool:
    _dict = {}
    for num in nums:
        if _dict.get(num, None) != None:
            return True
        _dict[num] = 0

    return False

containsDuplicate([1,2,3,4]), containsDuplicate([1,2,3,3])
```

```python
# https://leetcode.com/problems/missing-number/

def missingNumber(nums: List[int]) -> int:
    out = [-1] * (len(nums) + 1)
    for num in nums:
        out[num] = num

    for i in range(len(out)):
        if out[i] == -1:
            return i
        
missingNumber([3,0,1]), missingNumber([9,6,4,2,3,5,7,0,1])
```

```python
# https://leetcode.com/problems/find-all-numbers-disappeared-in-an-array/

def findDisappearedNumbers(nums: List[int]) -> List[int]:
    out = set(range(1, len(nums) + 1))
    for num in nums:
        if num in out:
            out.remove(num)
    return list(out)

findDisappearedNumbers([4,3,2,7,8,2,3,1]), findDisappearedNumbers([1,1])
```

```python
# https://leetcode.com/problems/single-number/

def singleNumber(nums: List[int]) -> int:
        nums.sort()
        for i in range(0, len(nums), 2):
            if i+1>=len(nums) or nums[i] != nums[i+1]:
                return nums[i]
            
singleNumber([2,2,1]), singleNumber([4,1,2,1,2])
```

```python
# https://leetcode.com/problems/convert-1d-array-into-2d-array/

def construct2DArray(original: List[int], m: int, n: int) -> List[List[int]]:
    out = []
    if m*n != len(original):
        return out
    return ([original[i:i+n] for i in range(0, len(original), n)])

construct2DArray([1,2,3,4], 2, 2), construct2DArray([1,2,3], 1, 3)
```

```python
# https://leetcode.com/problems/move-zeroes/

def moveZeroes(nums: List[int]):
    """
    Do not return anything, modify nums in-place instead.
    """
    position = 0
    zeros = 0
    for num in nums:
        if num != 0:
            nums[position] = num
            position += 1
        else:
            zeros += 1

    for i in range(len(nums) - 1, -1, -1):
        if zeros <= 0:
            break
        nums[i] = 0
        zeros -= 1
    return nums
        
moveZeroes([0,1,0,3,12]), moveZeroes([0])
```

```python
# https://leetcode.com/problems/product-of-array-except-self/

def productExceptSelf(nums: List[int]) -> List[int]:
    out = [1]*len(nums)

    prod = 1
    for i in range(len(nums)):
        out[i] *= prod
        prod *= nums[i]

    prod = 1
    for i in range(len(nums) - 1, -1, -1):
        out[i] *= prod
        prod *= nums[i]

    return out

productExceptSelf([1,2,3,4]), productExceptSelf([-1,1,0,-3,3])
```

```python
# https://leetcode.com/problems/find-all-duplicates-in-an-array/

def findDuplicates(nums: List[int]) -> List[int]:
        out = []

        for num in nums:
            index = abs(num) - 1
            if nums[index] < 0:
                out.append(abs(num))
            else:
                nums[index] *= -1
        
        return out
    
findDuplicates([4,3,2,7,8,2,3,1]), findDuplicates([1,1,2])
```

```python
# https://leetcode.com/problems/set-matrix-zeroes/

def setZeroes(matrix: List[List[int]]):
        """
        Do not return anything, modify matrix in-place instead.
        """
        row = set()
        column = set()

        for i in range(len(matrix)):
            for j in range(len(matrix[0])):
                if matrix[i][j] == 0:
                    row.add(i)
                    column.add(j)
        
        for i in range(len(matrix)):
            for j in range(len(matrix[0])):
                if i in row or j in column:
                    matrix[i][j] = 0
        
        return matrix
    
setZeroes([[1,1,1],[1,0,1],[1,1,1]]), setZeroes([[0,1,2,0],[3,4,5,2],[1,3,1,5]])
```

```python
# https://leetcode.com/problems/spiral-matrix/

def spiralOrder(matrix: List[List[int]]) -> List[int]:
    return matrix and [*matrix.pop(0)] + spiralOrder([*zip(*matrix)][::-1])

spiralOrder([[1,2,3],[4,5,6],[7,8,9]])
```

```python
# https://leetcode.com/problems/rotate-image/

def rotate(matrix: List[List[int]]) -> None:
    """
    Do not return anything, modify matrix in-place instead.
    """
    matrix[:] = zip(*matrix[::-1])

```
